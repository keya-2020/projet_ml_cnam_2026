```python
# ==========================================================
# TRAIN_TEST_SPLIT_TEMPORAL.PY
# ==========================================================

# ==========================================================
# IMPORTS
# ==========================================================
import pandas as pd

# ==========================================================
# LOAD DATA
# ==========================================================
df = pd.read_csv("data_processed.csv")

# ==========================================================
# TEMPORAL SPLIT
# ==========================================================

train_df = df[
    df["month"] <= 5
]

test_df = df[
    df["month"] >= 6
]

print("===================================================")
print("TEMPORAL SPLIT")
print("===================================================")

print(f"Train shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")

# ==========================================================
# FRAUD DISTRIBUTION
# ==========================================================

print("\nTrain Fraud Rate:")
print(train_df["fraud_bool"].mean())

print("\nTest Fraud Rate:")
print(test_df["fraud_bool"].mean())

# ==========================================================
# SAVE
# ==========================================================

train_df.to_csv(
    "train_temporal_dataprocessed.csv",
    index=False
)

test_df.to_csv(
    "test_temporal_dataprocessed.csv",
    index=False
)

print("\nTemporal split completed!")
```
